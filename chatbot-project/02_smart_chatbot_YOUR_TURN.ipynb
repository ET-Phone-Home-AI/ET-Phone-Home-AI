## STEP 5 — Describe YOUR Business ✏️

This is where **you** do the work — and it's all plain English, no coding.

Find the `BUSINESS_CONTEXT` section in the code cell below.
**Replace the example text** with your own business idea.

Your description should include:
- What your business does / sells
- Hours, location, pricing (make it up!)
- Any special rules or policies
- The personality you want your bot to have

**Ideas:** food truck, clothing brand, gym, nail salon, travel agency, game store, tutoring service...

> 💡 The more detail you write, the smarter your chatbot will be. This is called a **system prompt** —
> it's how businesses customize AI without writing any code.

---

```python
import groq
import gradio as gr
from google.colab import userdata

# Load your API key from Colab Secrets (safe — never exposed in code)
GROQ_API_KEY = userdata.get('GROQ_API_KEY')

# ==============================================================
# ✏️  EDIT THIS SECTION — Describe your business in plain English
# The AI will use this to answer any question a customer asks.
# ==============================================================

BUSINESS_CONTEXT = """
You are a helpful customer service assistant for Campus Bites,
a food truck at VCU in Richmond, Virginia.

About the business:
- We serve build-your-own rice bowls, tacos, and wraps
- Open Monday through Friday, 11am to 3pm, near the Compass
- Prices range from $8 to $13
- We have vegetarian, vegan, and gluten-free options clearly labeled
- We accept cash, card, and Venmo (@CampusBites)
- No pre-orders — first come, first served
- Follow us on Instagram @CampusBitesVCU for the daily special

Your personality:
- Friendly, casual, and quick — you know students are busy
- Use short answers, get to the point
- If you don't know something specific (like today's special),
  tell them to check Instagram — do NOT just say "call us"
"""

# ==============================================================
# The code below powers the smart chatbot — no need to change it
# ==============================================================

client = groq.Groq(api_key=GROQ_API_KEY)

def smart_chatbot(message, history):
    # Build the conversation history for the AI
    messages = [{"role": "system", "content": BUSINESS_CONTEXT}]

    for human_msg, ai_msg in history:
        messages.append({"role": "user",      "content": human_msg})
        messages.append({"role": "assistant", "content": ai_msg})

    messages.append({"role": "user", "content": message})

    try:
        response = client.chat.completions.create(
            model="llama-3.3-70b-versatile",
            messages=messages,
            max_tokens=300
        )
        return response.choices[0].message.content

    except Exception as e:
        if "rate_limit" in str(e).lower():
            return "I'm getting a lot of questions right now! Try again in a minute."
        return f"Something went wrong. Check that your API key is saved correctly in Colab Secrets."


# Launch the chat interface
gr.ChatInterface(
    smart_chatbot,
    title="🤖 My Business AI Assistant",
    description="Powered by Llama 3.3 via Groq — Ask me anything!"
).launch()
```
